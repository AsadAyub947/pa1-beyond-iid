```python
import os
import zipfile
from google.colab import drive

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Set up Drive persistence directory
DRIVE_PERSIST_DIR = "/content/drive/MyDrive/PA1_task4_results"
os.makedirs(f"{DRIVE_PERSIST_DIR}/checkpoints", exist_ok=True)
os.makedirs(f"{DRIVE_PERSIST_DIR}/results", exist_ok=True)

# 3. Path to your task4.zip in Google Drive
# (Change path if task4.zip is inside a subfolder, e.g., '/content/drive/MyDrive/PA1/task4.zip')
zip_path = '/content/drive/MyDrive/task4.zip'

print(f"Extracting {zip_path}...")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall('/content/')

# 4. Navigate into task4 directory
%cd /content/task4

# 5. Symlink checkpoints and results to Google Drive for automatic persistence
!rm -rf checkpoints results
!ln -s {DRIVE_PERSIST_DIR}/checkpoints checkpoints
!ln -s {DRIVE_PERSIST_DIR}/results results

print("\nSetup complete! Symlinks established to Google Drive:")
!ls -la checkpoints results
```
